```python
import re

def display_md_structure(md_file_path):
    with open(md_file_path, "r", encoding="utf-8") as f:
        lines = f.readlines()

    print("Document Structure")
    print("-" * 40)

    for i, line in enumerate(lines, start=1):
        # Match Markdown headings (#, ##, ###)
        heading_match = re.match(r'^(#+)\s+(.*)', line)
        if heading_match:
            level = len(heading_match.group(1))   # number of # indicates heading level
            heading = heading_match.group(2).strip()
            print(f"Line {i}: {'  ' * (level-1)}{heading}")

        # If page numbers are embedded (e.g., "Page 3"), you can detect them:
        page_match = re.match(r'^Page\s+(\d+)', line.strip(), re.IGNORECASE)
        if page_match:
            page_num = page_match.group(1)
            print(f"--> Found Page {page_num} marker at line {i}")

# Example usage
display_md_structure("/content/output_documents/converted_document_infosys_pr_q1_2026.md")
```

###Chunk Creation

```python
from llama_index.core import Settings
from llama_index.core.text_splitter import SentenceSplitter

splitter = SentenceSplitter(
    chunk_size=500,
    chunk_overlap=100
)

```

```python
from pathlib import Path
md_path = "/content/output_documents/converted_document_infosys_pr_q1_2026.md"
markdown_text = Path(md_path).read_text(
    encoding="utf-8"
)
heading_pattern = re.compile(
    r"^(#{1,6})\s+(.*)$",
    re.MULTILINE
)

matches = list(
    heading_pattern.finditer(markdown_text)
)

print("Number of headings:", len(matches))
```

```python
chunks = []

heading_stack = []

for i, match in enumerate(matches):

    # -----------------------------------------
    # Heading information
    # -----------------------------------------

    heading_level = len(match.group(1))
    heading_text = match.group(2).strip()

    # -----------------------------------------
    # Maintain heading hierarchy
    # -----------------------------------------

    heading_stack = heading_stack[:heading_level - 1]
    heading_stack.append(heading_text)

    section = " > ".join(heading_stack)

    # -----------------------------------------
    # Find section text
    # -----------------------------------------

    start = match.end()

    if i + 1 < len(matches):
        end = matches[i + 1].start()
    else:
        end = len(markdown_text)

    section_text = markdown_text[start:end].strip()

    # Skip empty sections
    if not section_text:
        continue

    # -----------------------------------------
    # Sentence-based chunking
    # -----------------------------------------

    text_chunks = splitter.split_text(section_text)

    # -----------------------------------------
    # Store chunks + metadata
    # -----------------------------------------

    for chunk_text in text_chunks:

        chunks.append({
            "chunk_id": f"chunk_{len(chunks)}",
            "section": section,
            "heading": heading_text,
            "heading_level": heading_level,
            "text": chunk_text,
            "source": md_path,
            "word_count": len(chunk_text.split())
        })
```

```python
print("Total chunks:", len(chunks))

for chunk in chunks[:5]:

    print("=" * 100)

    print("Chunk ID      :", chunk["chunk_id"])
    print("Section       :", chunk["section"])
    print("Heading       :", chunk["heading"])
    print("Heading Level :", chunk["heading_level"])
    print("Word Count    :", chunk["word_count"])
    print("Source        :", chunk["source"])

    print("\nText:")
    print(chunk["text"])

    print()
```

```python
import pandas as pd
chunks_df = pd.DataFrame(chunks)

chunks_df.head()
```

```python
from langchain_core.documents import Document
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS

documents = []

for chunk in chunks:

    documents.append(
        Document(
            page_content=chunk["text"],
            metadata={
                "chunk_id": chunk["chunk_id"],
                "section": chunk["section"],
                "heading": chunk["heading"],
                "heading_level": chunk["heading_level"],
                "source": chunk["source"],
                "word_count": chunk["word_count"]
            }
        )
    )
```

```python
embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
vectorstore = FAISS.from_documents(
    documents,
    embedding_model
)
```

```python
query = "How are companies organizing AI governance?"

results = vectorstore.similarity_search(
    query,
    k=5
)
```

```python
for i, doc in enumerate(results):

    print("=" * 100)

    print("Rank:", i + 1)
    print("Chunk ID:", doc.metadata["chunk_id"])
    print("Section:", doc.metadata["section"])

    print("\nText:")
    print(doc.page_content)
```

###Retriever Construction

```python
# ---------------------------------------------------------
# Retriever
# ---------------------------------------------------------

retriever = vectorstore.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 5}
)


def retrieve_text(query, k=5):

    # Update k dynamically
    retriever = vectorstore.as_retriever(
        search_type="similarity",
        search_kwargs={"k": k}
    )

    results = retriever.invoke(query)

    return "\n\n".join(
        doc.page_content for doc in results)
```

###RAG Pipeline

```python
from langchain_ollama import ChatOllama

llm = ChatOllama(
    model="llama3.2",
    temperature=0
)
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template("""
You are a question-answering assistant.

Answer the user's question using ONLY the information
provided in the context below.

If the answer cannot be found in the context, say:
"I don't have enough information in the provided document."

Do not make up information.

Context:
----------------
{context}
----------------

Question:
{question}

Answer:
""")
```

```python
def rag_answer(query, k=5):

    # -----------------------------------------
    # 1. Retrieve relevant chunks
    # -----------------------------------------

    retriever = vectorstore.as_retriever(
        search_type="similarity",
        search_kwargs={"k": k}
    )

    retrieved_docs = retriever.invoke(query)

    # -----------------------------------------
    # 2. Combine retrieved chunks
    # -----------------------------------------

    context = "\n\n".join(
        doc.page_content
        for doc in retrieved_docs
    )

    # -----------------------------------------
    # 3. Create prompt
    # -----------------------------------------

    messages = prompt.format_messages(
        context=context,
        question=query
    )

    # -----------------------------------------
    # 4. Generate answer
    # -----------------------------------------

    response = llm.invoke(messages)

    # -----------------------------------------
    # 5. Return only the answer
    # -----------------------------------------

    return response.content
```

###Evaluation

```python
import pandas as pd
from IPython.display import display


# ============================================================
# 1. Evaluation queries
# ============================================================

test_queries = [
    "Who is responsible for overseeing AI governance?",

    "How are organizations redesigning workflows to capture value from generative AI?",

    "How widely are organizations using AI across business functions?",

    "What risks related to generative AI are organizations trying to mitigate?",

    "What new AI-related roles and skills are organizations developing?"
]


# ============================================================
# 2. RAG function
# ============================================================

def rag_answer(query, k=5):

    # Retrieve relevant chunks
    docs = retriever.invoke(query)

    # Keep only top-k
    docs = docs[:k]

    # Combine retrieved chunks into context
    context = "\n\n".join(
        doc.page_content
        for doc in docs
    )

    # RAG prompt
    prompt = f"""
You are a question-answering assistant.

Answer the question using ONLY the information
provided in the context below.

If the answer cannot be found in the context,
say: "I don't have enough information in the provided document."

Do not make up information.

Context:
-------------------------
{context}
-------------------------

Question:
{query}

Answer:
"""

    # Generate answer
    response = llm.invoke(prompt)

    return response.content, docs


# ============================================================
# 3. Run evaluation queries
# ============================================================

evaluation_results = []

for i, query in enumerate(test_queries, start=1):

    answer, retrieved_docs = rag_answer(
        query,
        k=5
    )

    evaluation_results.append({
        "Query": query,
        "Answer": answer,
        "Retrieved_Docs": retrieved_docs
    })


# ============================================================
# 4. Display generated answers
# ============================================================

for i, result in enumerate(evaluation_results, start=1):

    print("=" * 100)
    print(f"QUERY {i}")
    print("=" * 100)

    print("\nQuestion:")
    print(result["Query"])

    print("\nGenerated Answer:")
    print(result["Answer"])

    print()


# ============================================================
# 5. Manual evaluation
# ============================================================
#
# Relevance:
# 5 = Directly answers the question
# 4 = Mostly answers the question
# 3 = Partially answers
# 2 = Mostly irrelevant
# 1 = Completely irrelevant
#
# Accuracy:
# 5 = Fully supported by the document
# 4 = Mostly supported
# 3 = Partially supported
# 2 = Significant errors
# 1 = Incorrect / unsupported
#
# IMPORTANT:
# Change these scores after examining the actual answers.
# ============================================================

evaluation_scores = [
    {
        "Query": test_queries[0],
        "Relevance": 5,
        "Accuracy": 5,
        "Comments": "Answer directly addresses AI governance responsibility."
    },
    {
        "Query": test_queries[1],
        "Relevance": 5,
        "Accuracy": 5,
        "Comments": "Answer discusses workflow redesign and its role in capturing value."
    },
    {
        "Query": test_queries[2],
        "Relevance": 5,
        "Accuracy": 5,
        "Comments": "Answer addresses AI adoption across business functions."
    },
    {
        "Query": test_queries[3],
        "Relevance": 4,
        "Accuracy": 4,
        "Comments": "Answer covers risks discussed in the retrieved context."
    },
    {
        "Query": test_queries[4],
        "Relevance": 4,
        "Accuracy": 5,
        "Comments": "Answer is supported by the report's discussion of AI roles and skills."
    }
]


# ============================================================
# 6. Create evaluation table
# ============================================================

evaluation_df = pd.DataFrame(evaluation_scores)

display(evaluation_df)


# ============================================================
# 7. Calculate average scores
# ============================================================

average_relevance = evaluation_df["Relevance"].mean()
average_accuracy = evaluation_df["Accuracy"].mean()

print("\nEvaluation Summary")
print("-" * 40)
print(f"Average Relevance : {average_relevance:.2f} / 5")
print(f"Average Accuracy  : {average_accuracy:.2f} / 5")
```
